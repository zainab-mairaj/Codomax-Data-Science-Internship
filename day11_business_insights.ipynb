**Prepared by:** Zainab Mairaj
**Objective:** Synthesize key quantitative findings into actionable business insights and strategic observations. 

## 🔍 Key Observations & Business Insights

### 1. Demand & Volume: Class-Based Passenger Distribution
- **Observation:** Class 3 (Economy) accounted for over 50% of total passengers, outnumbering Class 1 and Class 2 combined.
- **Insight:** Primary market revenue volume was driven by low-cost ticket sales, showing high price sensitivity among customers.

### 2. Operational Risk: Overall Survival Rate
- **Observation:** The overall survival rate was approximately **38.4%** (over 61% mortality).
- **Insight:** Demonstrates critical deficiencies in emergency response protocols and safety capacity relative to passenger volume.

### 3. Target Demographics: Age Distribution
- **Observation:** Passenger age distribution heavily skews toward young adults aged **20–35 years**.
- **Insight:** The core consumer demographic consisted of working-age individuals and young families.

### 4. Pricing Strategy: High Fare Variance
- **Observation:** Fares ranged from **$0.00 to $512.33**, with an average ticket fare of **$32.20**.
- **Insight:** Multi-tier pricing successfully captured both high-margin luxury segments and high-volume budget segments.

### 5. Equity & Crisis Management: Survival Bias
- **Observation:** Higher-class passengers and vulnerable demographics experienced higher survival proportions.
- **Insight:** Highlights structural gaps in evacuation access and safety priority management during critical incidents.
