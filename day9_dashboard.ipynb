# Day 9: Titanic Data Analysis Dashboard

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset load karna
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Dashboard Environment Loaded Successfully!")
```

```python
# --- 1. KPI / Summary Metrics ---
total_passengers = len(df)
avg_fare = df['Fare'].mean()
survival_rate = (df['Survived'].sum() / total_passengers) * 100

print("==========================================")
print("       🚢 TITANIC EXECUTIVE DASHBOARD     ")
print("==========================================")
print(f" Total Passengers  : {total_passengers}")
print(f" Average Ticket    : ${avg_fare:.2f}")
print(f" Survival Rate     : {survival_rate:.1f}%")
print("==========================================\n")

# --- 2. Dashboard Visual Grid Setup (2 rows, 2 columns) ---
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Titanic Dataset Key Metrics & Visual Analytics', fontsize=16, fontweight='bold')

# Chart 1: Bar Chart (Passenger Count by Class)
class_counts = df['Pclass'].value_counts()
axes[0, 0].bar(class_counts.index.astype(str), class_counts.values, color=['#3498db', '#e67e22', '#2ecc71'])
axes[0, 0].set_title('Passenger Distribution by Class')
axes[0, 0].set_xlabel('Class')
axes[0, 0].set_ylabel('Count')
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.6)

# Chart 2: Pie Chart (Survival Breakdown)
survived_counts = df['Survived'].value_counts()
axes[0, 1].pie(survived_counts.values, labels=['Died', 'Survived'], autopct='%1.1f%%', colors=['#e74c3c', '#2ecc71'], startangle=90)
axes[0, 1].set_title('Survival Percentage')

# Chart 3: Line Chart (Fare Trend for Sample Passengers)
sample_df = df.head(50)
axes[1, 0].plot(sample_df.index, sample_df['Fare'], color='#9b59b6', marker='o', linewidth=1.5, markersize=3)
axes[1, 0].set_title('Ticket Fare Trend (First 50 Passengers)')
axes[1, 0].set_xlabel('Passenger Index')
axes[1, 0].set_ylabel('Fare ($)')
axes[1, 0].grid(True, linestyle=':', alpha=0.6)

# Chart 4: Histogram/Bar (Age Distribution)
axes[1, 1].hist(df['Age'].dropna(), bins=15, color='#34495e', edgecolor='white')
axes[1, 1].set_title('Passenger Age Distribution')
axes[1, 1].set_xlabel('Age')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].grid(axis='y', linestyle='--', alpha=0.6)

# Layout polish
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
