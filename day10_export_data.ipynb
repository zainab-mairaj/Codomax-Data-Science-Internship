# Day 10: Data Exporting

```python
import pandas as pd


url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)


df['Age'] = df['Age'].fillna(df['Age'].median())  # Fill missing age with median
df = df.drop(columns=['Cabin'])                  # Drop column with too many missing values
df = df.dropna()                                 # Drop remaining missing rows

# 3. Export Cleaned Dataset to CSV
df.to_csv('cleaned_titanic_data.csv', index=False)

print("✅ Success! Cleaned dataset saved as 'cleaned_titanic_data.csv'")
print(f"Total Rows Saved: {len(df)}")
```
